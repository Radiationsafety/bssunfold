# Example for bssunfold package with Genetic solvers in mealpy

- 8 MEALPY solvers: `pso` (chaotic PSO, default), `ga`, `de`, `es`, `ep`,
    `abc`, `gwo`, `cmaes`


    `unfold_genetic()` method
  using population-based meta-heuristic algorithms from MEALPY. Minimizes
  `||A·x − b||²/||b||² + α·||x||_norm` with `x ≥ 0` and optional
  second-difference smoothing and Shannon-entropy terms, following the
  PSO (Shahabinejad & Sohrabpour 2017), GA (Suman & Sarkar 2012) and
  entropy-based (Woo et al. 2019) unfolding works. No initial spectrum is
  required (random population initialization).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from bssunfold import Detector, RF_LANL
from bssunfold.utils.plotting import plot_comparison
det = Detector(RF_LANL)
```

## Get effective readings for spectra

```python
reference_spectrum = pd.read_csv('../tests/MonteCarlo_Calculated_spectra_from_IAEA_Comp_for_comparison.csv')
readings = det.get_effective_readings_for_spectra(reference_spectrum[['E_MeV','ISO_ref_Cf252']])
readings
```

```python
results = {
    "genetic[pso-epochs=5000]": det.unfold_genetic(
        readings,
        regularization=2e-2,
        epoch=5000,
        smoothness_order=0,
        pop_size=1000,
        solver="pso",
    ),
    "genetic[ga]": det.unfold_genetic(
        readings,
        regularization=2e-4,
        epoch=2000,
        smoothness_order=1,
        entropy_weight=0.5,
        pop_size=20,
        solver="ga",
    ),
    "genetic[de]": det.unfold_genetic(
        readings,
        regularization=2e-4,
        epoch=4000,
        smoothness_order=1,
        entropy_weight=0.5,
        pop_size=20,
        solver="de",
    ),
    "genetic[es]": det.unfold_genetic(
        readings,
        regularization=2e-4,
        solver="es",
    ),
    "genetic[ep]": det.unfold_genetic(
        readings,
        regularization=2e-4,
        solver="ep",
    ),
    "genetic[abc]": det.unfold_genetic(
        readings,
        regularization=2e-4,
        solver="abc",
    ),
    "genetic[gwo]": det.unfold_genetic(
        readings,
        regularization=2e-4,
        solver="gwo",
    ),
    "genetic[cmaes]": det.unfold_genetic(
        readings,
        regularization=1e-4,
        solver="cmaes",
    ),
}

fig, ax = plot_comparison(
    results=results,
    readings=readings,
    reference_spectrum=reference_spectrum[["E_MeV", "ISO_ref_Cf252"]],
)
```
